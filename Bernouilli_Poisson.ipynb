```python
#FUNCIONES DE DISTRIBUCIÓN
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
from statsmodels.distributions.empirical_distribution import ECDF

np.random.seed(0)
sample = np.random.normal(loc=mu,scale=sigma,size=n)

# CREAMOS LA FUNCIÓN DE DISTRIBUCIÓN EMPÍRICA
ecdf = ECDF(sample)

#Generamos valores alrededor de la media, el dominio son los puntos en los que vamos a evaluar la funcion de distribucion, es decir, F(X<dominio[i])
mu = 50
sigma = 1
n=100

dominio=np.arange(start=mu-3*sigma,stop=mu+3*sigma,step=0.05)
long_dominio=len(dominio)
sample=np.random.normal(loc=mu,scale=sigma,size=n)

func_distr_emp=np.zeros(long_dominio)
for i in range(long_dominio):
    mask=sample<dominio[i]
    count=len(sample[mask])
    val=count/n
    func_dist_emp[i]=val
```

```python
#REPRESENTAMOS GRÁFICAMENTE LA FUNCIÓN DE DISTRIBUCIÓN EMPÍRICA

plt.figure(figsize=(8,5)) #Definimos el tamaño del gráfico.

# REPRESENTAMOS LA MUESTRA
plt.plot(sample, np.zeros_like(sample),'x',color='black')

#DIBUJAMOS LA FUNCIÓN DE DISTRIBUCIÓN EMPÍRICA
plt.plot(dominio,func_distr_emp,label ="Función de distribución empírica a mano")

#PARA COMPARAR, DIBUJAMOS LAS FUNCIONES DE DISTRIBUCIÓN(cdf) Y DENSIDAD(pdf) TEÓRICAS
plt.plot(ecdf.x, stats.norm.cdf(x=ecdf.x, loc=mu, scale = sigma),label = 'Función de distribución analítica N(50,1)')
plt.plot(ecdf.x, stats.norm.pdf(x=ecdf.x, loc=mu, scale = sigma),'--',label= 'Función de densidad analítica N(50,1)')

#Configuramos el gráfico
plt.legend()
plt.title("Comparativa")
plt.show()
```
